```python
from langgraph.store.memory import InMemoryStore
from langchain_huggingface import HuggingFaceEmbeddings
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
store = InMemoryStore()
```

```python
namespace1 = ("user", "u1")
```

```python
store.put(namespace1, "1", {"data": "User likes pizza"})
store.put(namespace1, "2", {"data": "User prefers dark mode"})
```

```python
namespace2 = ("user", "u2")

# Add memories
store.put(namespace2, "1", {"data": "User likes pasta"})
store.put(namespace2, "2", {"data": "User prefers grid style navigation"})
```

```python
store.get(namespace2, "2")
```

```python
items = store.search(namespace2)

for item in items:
    print(item.value)
```

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
store = InMemoryStore(index= {'embed': embeddings, 'dims': 728})
```

```python
store.put(namespace1, "1", {"data": "User prefers concise answers over long explanations"})
store.put(namespace1, "2", {"data": "User likes examples in Python"})
store.put(namespace1, "3", {"data": "User usually works late at night"})
store.put(namespace1, "4", {"data": "User prefers dark mode in applications"})
store.put(namespace1, "5", {"data": "User is learning machine learning"})
store.put(namespace1, "6", {"data": "User dislikes overly theoretical explanations"})
store.put(namespace1, "7", {"data": "User prefers step-by-step reasoning"})
store.put(namespace1, "8", {"data": "User is based in India"})
store.put(namespace1, "9", {"data": "User likes real-world analogies"})
store.put(namespace1, "10", {"data": "User prefers bullet points over paragraphs"})
```

```python
items = store.search(namespace1, query="what is the user currently learning", limit=1)

for item in items:
    print(item.value)
```

```python
items = store.search(namespace1, query="what are user's preferences", limit=3)

for item in items:
    print(item.value)
```
